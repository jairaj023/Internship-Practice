```python
import nltk
import spacy

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer, PorterStemmer
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
text = "We are looking for candidates with strong experience in Python and SQL."

print("=" * 60)
print("NLP TEXT PREPROCESSING")
print("=" * 60)
```

```python
# 1. Tokenization
tokens = word_tokenize(text)

print("\n1. TOKENIZATION")
print(tokens)
```

```python
# 2. Lowercasing
lowercase_text = text.lower()

print("\n2. LOWERCASE")
print(lowercase_text)
```

```python
# 3. Stopword Removal
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in tokens
    if word.lower() not in stop_words and word.isalpha()
]

print("\n3. STOPWORD REMOVAL")
print(filtered_words)
```

```python
# 4. Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word.lower())
    for word in filtered_words
]

print("\n4. LEMMATIZATION")
print(lemmatized_words)
```

```python
# 5. Stemming
stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word.lower())
    for word in filtered_words
]

print("\n5. STEMMING")
print(stemmed_words)
```

```python
# Stemming vs Lemmatization
print("\nSTEMMING VS LEMMATIZATION")
print("-" * 40)
print(f"{'Word':<15}{'Lemma':<15}{'Stem'}")
print("-" * 40)

for word in filtered_words:
    lemma = lemmatizer.lemmatize(word.lower())
    stem = stemmer.stem(word.lower())
    print(f"{word:<15}{lemma:<15}{stem}")
```

```python
# 6. Sentence Segmentation
text_sentences = """
We are looking for candidates with strong experience in Python and SQL.
The candidate should also have good analytical skills.
Knowledge of Power BI is an advantage.
"""

sentences = sent_tokenize(text_sentences)

print("\n6. SENTENCE SEGMENTATION")

for i, sentence in enumerate(sentences, start=1):
    print(f"Sentence {i}: {sentence.strip()}")

print("\n" + "=" * 51)
print("PREPROCESSING COMPLETED")
print("=" * 51)
```
