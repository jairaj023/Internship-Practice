```python
df = pd.read_csv('raw_jobs.csv')
```

```python
df.head()
```

```python
skills = ["python", "sql", "power bi", "tableau", "excel", "aws"]
```

```python
def extract_skills(text):
    found = []

    # Handle missing values
    if pd.isna(text):
        return found

    # Convert text to lowercase
    text = str(text).lower()

    # Search for each skill
    for skill in skills:
        if skill in text:
            found.append(skill)

    return found
```

```python
df["extracted_skills"] = df["job_description"].apply(extract_skills)
```

```python
print(df[['job_id', 'job_title', 'job_description', 'extracted_skills']].head(10))
```

```python
print("\nExtracted Skills:")
print(df["extracted_skills"].head(10))
```

```python
for skill in skills:
    df[skill] = df["job_description"].fillna("").str.lower().str.contains(skill, regex=False)
```

```python
print(df[["job_id", "job_title", "python", "sql", "power bi", "tableau", "excel", "aws"]].head(10))
```

```python
df["extracted_skills"] = df["extracted_skills"].apply(lambda x: ", ".join(x) if x else "No skills found")

print(df[["job_id", "job_title", "extracted_skills"]].head(10))
```
