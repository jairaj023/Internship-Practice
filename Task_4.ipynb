```python
print("Missing Job Descriptions : ", df['job_description'].isna().sum())
```

```python
import re
def clean_text(text):
  text = str(text)

  text = re.sub(r"<.*?", " ", text)
  text = re.sub(r"http\S+", "", text)
  text = re.sub(r"\S+@\S+", "", text)
  text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)
  text = re.sub(r"\s+", " ", text)
  text = text.strip().lower()

  return text
```

```python
df['clean_description'] = df['job_description'].apply(clean_text)
```

```python
df[['clean_description', 'job_description']].head(10)
```

```python
df = df.drop_duplicates(subset = ['clean_description'])
```

```python
df = df[df['clean_description'].str.strip() != ""]
```

```python
df = df.reset_index(drop = True)
```

```python
df.to_csv('clean_job.csv', index = True)
```
