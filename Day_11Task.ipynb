```python
import pandas as pd
from sentence_transformers import SentenceTransformer, util
```

```python
df = pd.read_csv("raw_jobs.csv")
```

```python
print(df.head())
```

```python
print(df.columns)
```

```python
text_column = "job_description"
```

```python
print(df[text_column].head())
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
text1 = "ML"
text2 = "Machine Learning"

embedding1 = model.encode(text1)
embedding2 = model.encode(text2)
```

```python
similarity = util.cos_sim(embedding1, embedding2)

print("Similarity:", similarity.item())
```

```python
skills = [
    "Python",
    "Machine Learning",
    "Deep Learning",
    "Natural Language Processing",
    "Data Science",
    "Power BI",
    "SQL",
    "MySQL",
    "PostgreSQL",
    "Excel",
    "Java",
    "JavaScript",
    "React",
    "TensorFlow",
    "PyTorch"
]
```

```python
skill_embeddings = model.encode(skills, convert_to_tensor=True)

print("Skill embeddings created!")
```

```python
def extract_skills(text):

    text_embedding = model.encode(text, convert_to_tensor=True)

    similarities = util.cos_sim(text_embedding, skill_embeddings)[0]

    found_skills = []

    for i, score in enumerate(similarities):

        if score >= 0.30:
            found_skills.append(skills[i])

    return found_skills
```

```python
text = """
We are looking for a data scientist with experience in ML,
Python, deep learning and SQL.
"""

result = extract_skills(text)

print("Skills found:")
print(result)
```

```python
df["semantic_skills"] = df[text_column].apply(extract_skills)
```

```python
print(df[[text_column, "semantic_skills"]].head(10))
```

```python
df["semantic_skills"] = df["semantic_skills"].apply(lambda x: ", ".join(x))
```

```python
print(df[[text_column, "semantic_skills"]].head())
```

```python
examples = [
    "ML",
    "Machine Learning",
    "deep neural networks",
    "NLP",
    "data analysis using Python",
    "database management using SQL"
]

for text in examples:

    result = extract_skills(text)

    print("\nText:", text)
    print("Skills:", result)
```

```python
pairs = [
    ("ML", "Machine Learning"),
    ("NLP", "Natural Language Processing"),
    ("AI", "Artificial Intelligence"),
    ("Python programming", "Python"),
    ("SQL database", "SQL"),
    ("Excel spreadsheet", "Excel")
]

for text1, text2 in pairs:

    emb1 = model.encode(text1)
    emb2 = model.encode(text2)

    score = util.cos_sim(emb1, emb2).item()

    print(text1, "<->", text2, "=", round(score, 3))
```

```python
df.to_csv("semantic_skill_extracted.csv", index=False)

print("\nFile saved successfully!")
```
