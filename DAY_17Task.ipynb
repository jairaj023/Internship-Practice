Alias Dictionary

```python
skill_aliases = {

    "python": "Python",
    "python3": "Python",
    "python 3": "Python",
    "python programming": "Python",

    "powerbi": "Power BI",
    "power bi": "Power BI",
    "power-bi": "Power BI",

    "postgres": "PostgreSQL",
    "postgresql": "PostgreSQL",
    "postgre sql": "PostgreSQL",

    "javascript": "JavaScript",
    "java script": "JavaScript",
    "js": "JavaScript",

    "react": "React",
    "reactjs": "React",
    "react js": "React",

    "node": "Node.js",
    "nodejs": "Node.js",
    "node js": "Node.js",

    "mysql": "MySQL",
    "my sql": "MySQL",

    "sql": "SQL",

    "machine learning": "Machine Learning",
    "machinelearning": "Machine Learning",
    "ml": "Machine Learning",

    "artificial intelligence": "Artificial Intelligence",
    "artificialintelligence": "Artificial Intelligence",
    "ai": "Artificial Intelligence",

    "excel": "Excel",
    "microsoft excel": "Excel",
    "ms excel": "Excel",

    "tableau": "Tableau",

    "git": "Git",

    "github": "GitHub",
    "git hub": "GitHub",

    "html": "HTML",
    "html5": "HTML",

    "css": "CSS",
    "css3": "CSS",

    "java": "Java",

    "c++": "C++",
    "cpp": "C++",
    "c plus plus": "C++"
}
```

Function to Normalize One Skill

```python
def normalize_one_skill(skill):

    # Convert to string
    skill = str(skill)

    # Remove extra spaces
    skill = skill.strip()

    # Lowercase
    skill = skill.lower()

    # Remove special characters
    skill = re.sub(r"[^a-z0-9+#. ]", "", skill)

    # Remove extra spaces
    skill = re.sub(r"\s+", " ", skill)

    # Alias matching
    if skill in skill_aliases:
        return skill_aliases[skill]

    # If alias doesn't exist
    return skill.title()
```

Normalize Extracted Skills

```python
df["Normalized Skills"] = df["extracted_skills"].apply(normalize_one_skill)
```

Compare Before and After

```python
print("\nBefore vs After Normalization:")

print(df[["extracted_skills", "Normalized Skills"]].head(20))
```

Number of Unique Skills

```python
print("\nUnique skills BEFORE normalization:")

print(df["extracted_skills"].nunique())

print("\nUnique skills AFTER normalization:")

print(df["Normalized Skills"].nunique())
```
