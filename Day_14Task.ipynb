Custom Skill Extraction

```python
skills = [
          "Python",
          "Java",
          "C++",
          "C#",
          "JavaScript",
          "SQL",
          "MySQL",
          "PostgreSQL",
          "MongoDB",
          "Oracle",
          "AWS",
          "Azure",
          "Google Cloud",
          "GCP",
          "Power BI",
          "Tableau",
          "Excel",
          "Python",
          "Pandas",
          "Numpy",
          "Scikit-learn",
          "TensorFlow",
          "PyTorch",
          "Machine Learning",
          "Data Science",
          "React",
          "Angular",
          "Node.js",
          "Django",
          "Flask",
          "HTML",
          "CSS",
          "Docker",
          "Kubernetes",
          "Git",
          "GitHub"
          ]
```

Create a skill extraction function

```python
def extract_skills(text, skills_list):

    text = str(text).lower()

    found_skills = []

    for skill in skills_list:
        if skill.lower() in text:
            found_skills.append(skill)

    return list(set(found_skills))
```

Extract skills from your dataset

```python
df["extracted_skills"] = df['job_description'].apply(lambda x: extract_skills(x, skills))

display(df[['job_title', 'job_description', 'extracted_skills']].head(10))
```

Create a function for transformer entity extraction

```python
def extract_entities(text):

    try:
        results = ner_model(text[:512])

        entities = []

        for entity in results:
            entities.append({
                "word": entity["word"],
                "entity": entity["entity_group"],
                "score": round(float(entity["score"]), 3)
            })

        return entities

    except Exception as e:
        return []
```

Test on one job description

```python
sample_text = df["job_description"].iloc[0]

entities = extract_entities(sample_text)

print("Job Description:\n")
print(sample_text)

print("\nExtracted Entities:\n")

for entity in entities:
    print(
        f"{entity['word']} → "
        f"{entity['entity']} "
        f"(Score: {entity['score']})"
    )
```

Comaparison function

```python
def compare_extraction(text):

  print("TEXT:")
  print(text)

  print("\n")
  print("PRETRAINED BERT NER RESULTS")
  print("\n")

  bert_results = extract_entities(text)

  if bert_results:
      for entity in bert_results:
          print(
              f"{entity['word']} → "
              f"{entity['entity']} "
              f"(Score: {entity['score']})"
          )
  else:
      print("No entities found")

  print("\n")
  print("CUSTOM SKILL EXTRACTION")
  print("\n")

  extracted = extract_skills(text, skills)

  if extracted:
    for skill in extracted:
      print(f"{skill} → SKILL")

  else:
    print("No skills found")
```

Test the comparison

```python
test_text = """We are looking for a Data Analyst with strong experience in Python, SQL, Power BI, Tableau, AWS and Excel."""

compare_extraction(test_text)
```

Apply to the complete dataset

```python
df_test = df.head(200).copy()

df_test["transformer_entities"] = df_test['job_description'].apply(lambda x: extract_entities(x))

print("Extraction completed!")
```
